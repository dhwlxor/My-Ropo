```python
import plotly.graph_objs as go

branches=['CSE','Mech','Electronics']

fy=[23,17,35]
sy=[20,23,30]
ty=[30,20,15]

trace1=go.Bar(x=branches,y=fy,name='FY')
trace2=go.Bar(x=branches,y=sy,name='SY')
trace3=go.Bar(x=branches,y=ty,name='TY')

data=[trace1, trace2, trace3]
layout=go.Layout(barmode='group', title='Departments')
fig=go.Figure(data=data, layout=layout)
fig.show()

data=[trace1, trace2, trace3]
layout=go.Layout(barmode='stack', title='Departments')
fig=go.Figure(data=data, layout=layout)
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

countries=["US","China","European Union","Russian Federation","Brazil","India","Rest of World"]

ghg=[16,15,12,6,5,4,42]
co2=[27,11,25,8,1,3,25]

fig=make_subplots(rows=1, cols=2, specs=[[{'type':'domain'},{'type':'domain'}]])

# 첫 번째 도넛 차트 (GHG)
fig.add_trace(go.Pie(labels=countries, values=ghg, name="GHG", hole=0.4), row=1, col=1)

# 두 번째 도넛 차트 (CO₂)
fig.add_trace(go.Pie(labels=countries, values=co2, name="CO₂", hole=0.4), row=1, col=2)

# 전체 레이아웃 설정
fig.update_layout(
    title_text="Global Emissions by Country",
    annotations=[dict(text='GHG', x=0.18, y=0.5, font_size=14, showarrow=False),
                 dict(text='CO₂', x=0.82, y=0.5, font_size=14, showarrow=False)]
)

fig.show()

```

```python
from sklearn.cluster import KMeans
import numpy as np

X = np.array([[2, 10], [2, 5], [8, 4], [5, 8], [7, 5], [6, 4], [1, 2], [4, 9]])

kmeans = KMeans(n_clusters=3).fit(X)

print("Labels: ", kmeans.labels_)
print("Cluster Centers: ", kmeans.cluster_centers_)
print("Predict Values: ", kmeans.predict([[1, 1]]))
```
